```python
# ISA Atmosphere Table
# Exercise 1

# Constants
T0 = 288.15          # Sea-level temperature [K] (15 °C)
P0 = 101325.0        # Sea-level pressure [Pa]
R = 287.0            # Specific gas constant for air [J/(kg·K)]
g = 9.81             # Gravitational acceleration [m/s²]
L = -0.0065          # Tropospheric lapse rate [K/m]
H_TROPOPAUSE = 11000 # Tropopause altitude [m]


def temperature(h):
    """Calculate ISA temperature at altitude h [m]."""
    if h <= H_TROPOPAUSE:
        return T0 + L * h
    else:
        return T0 + L * H_TROPOPAUSE


def pressure(h):
    """Calculate ISA pressure at altitude h [m]."""
    if h <= H_TROPOPAUSE:
        T = temperature(h)
        return P0 * (T / T0) ** (-g / (R * L))
    else:
        T11 = temperature(H_TROPOPAUSE)
        P11 = pressure(H_TROPOPAUSE)

        return P11 * (
            2.718281828459045 ** (
                -g * (h - H_TROPOPAUSE) / (R * T11)
            )
        )


def density(h):
    """Calculate ISA density at altitude h [m]."""
    T = temperature(h)
    P = pressure(h)

    return P / (R * T)


# Create table header
header = (
    f"{'Altitude (m)':>12}"
    f"{'Temperature (K)':>18}"
    f"{'Temperature (°C)':>20}"
    f"{'Pressure (Pa)':>18}"
    f"{'Density (kg/m³)':>20}"
)

separator = "-" * len(header)

# Store table lines
table = [header, separator]

# Calculate values from 0 to 20,000 m
for altitude in range(0, 20001, 1000):
    T = temperature(altitude)
    P = pressure(altitude)
    rho = density(altitude)

    line = (
        f"{altitude:>12}"
        f"{T:>18.2f}"
        f"{T - 273.15:>20.2f}"
        f"{P:>18.2f}"
        f"{rho:>20.4f}"
    )

    table.append(line)


# Print table to terminal
print("\n".join(table))

# Save table to a text file
with open("isa_atmosphere.txt", "w") as file:
    file.write("\n".join(table))

print("\nTable saved to isa_atmosphere.txt")
```
